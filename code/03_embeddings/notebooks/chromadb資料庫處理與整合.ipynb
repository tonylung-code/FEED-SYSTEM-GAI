```python
#資料庫 collection 名稱檢查
from pathlib import Path
import chromadb

DB_PATH = (Path.cwd().resolve().parents[2] / "databases" / "hiwin_vector_db").resolve()

# 1. 連接到目前主資料庫資料夾
client = chromadb.PersistentClient(path=str(DB_PATH))

# 2. 列出該資料夾下所有的 Collection 名稱
collections = client.list_collections()

print(f"--- 目前資料庫中的所有 Collection ({DB_PATH}) ---")
for col in collections:
    print(f"名稱: {col.name}")
```

```python
#資料庫內容檢查
import chromadb

collection_name = "screw_specs"

# 連接目前主資料庫
client = chromadb.PersistentClient(path=str(DB_PATH))
spec_collection = client.get_collection(name=collection_name)

# 抓取 5 筆資料來檢查
results = spec_collection.peek(limit=5)

print(f"--- 資料庫內容檢查 ({collection_name}) ---")
for i in range(len(results['ids'])):
    print(f"ID: {results['ids'][i]}")
    print(f"內容 (Document): {results['documents'][i][:80]}...")
    print(f"標籤 (Metadata): {results['metadatas'][i]}")
    print("-" * 30)
```
