```python
#資料庫 collection 名稱檢查
import chromadb

# 1. 連接到你的資料庫資料夾
client = chromadb.PersistentClient(path="./hiwin_vector_db")

# 2. 列出該資料夾下所有的 Collection 名稱
collections = client.list_collections()

print("--- 目前資料庫中的所有 Collection ---")
for col in collections:
    print(f"名稱: {col.name}")
```

```python
#資料庫內容檢查
import chromadb

# 連接現有資料庫
client = chromadb.PersistentClient(path="./hiwin_vector_db")
spec_collection = client.get_collection(name="hiwin_specs")

# 隨機抓取 5 筆資料來檢查
results = spec_collection.peek(limit=5)

print("--- 資料庫內容檢查 ---")
for i in range(len(results['ids'])):
    print(f"ID: {results['ids'][i]}")
    print(f"內容 (Document): {results['documents'][i][:50]}...") # 只看前50字
    print(f"標籤 (Metadata): {results['metadatas'][i]}")
    print("-" * 30)
```
